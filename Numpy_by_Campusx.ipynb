BROADCASTING
THE TERM BROADCASTING DESCRIBES HOW NUMPY TREATS ARRAY WITH DIFFERENT SHAPES DURING ARITHMETIC OPERATIONS. THE SMALLER ARRAY IS "BROADCAST" ACROSS THE LARGER ARRAY SO THAT THEY HAVE COMPATIBLE SHAPES

```python
#  same shape
a=np.arange(6).reshape(2,3)
b=np.arange(6,12).reshape(2,3)
print(a)
print(b)
print(a+b)
```

```python
# diff shape
a=np.arange(6).reshape(2,3)
b=np.arange(3).reshape(1,3)
print(a)
print(b)
print(a+b)
```

Broadcasting Rules
1)- Make the two arrays have the same number of dimensions.

    *)- if the numbers of dimensions of the two arrays are different , add new dimensions with size 1 to be head of the array with the smaller dimension .

2)- Make each dimension of the two arrays the same size

    *)- if the size of each dimension of the two arrays do not match , dimensions with size 1 are stretched to the size of other array.

    *)- if there is a dimension whose size is not 1 either of the two arrays , it cannot be broadcasted , and an error is raised .

```python
# more examples
import numpy as np
a = np.arange(12).reshape(4,3)
b = np.arange(3)
print(a)
print(b)

print(a+b)
```

```python
a = np.arange(3).reshape(1,3)
b = np.arange(3).reshape(3,1)

print(a)
print(b)

print(a+b)
```

```python
a=np.arange(3).reshape(1,3)
b=np.arange(4).reshape(4,1)
print(a)
print(b)
print(a+b)
```

```python
a=np.array([1])
# shape -> (1,1)
b=np.arange(4).reshape(2,2)
# shape -> (2,2)
print(a)
print(b)
print(a+b)
```

```python

def mse(actual,predicted):
  return np.mean((actual-predicted)**2)
mse(actual,predicted)
```

```python
# binary cross entropy
import numpy as np

y_true = 1
y_pred = 0.9

bce = -(y_true * np.log(y_pred) + (1 - y_true) * np.log(1 - y_pred))

print("Binary Cross Entropy:", bce)
```
